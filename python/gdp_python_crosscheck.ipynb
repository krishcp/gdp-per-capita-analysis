```python
import pandas as pd
import numpy as np
from scipy import stats

# Load your CSV (make sure the filename matches what you uploaded)
df = pd.read_csv('gdp_life_exp_final.csv')

# Pull out 2023 GDP and Life Expectancy columns
gdp = df['X2023..YR2023..x']
life = df['X2023..YR2023..y']

# Drop rows with missing values, same approach as in R
data = df.dropna(subset=['X2023..YR2023..x', 'X2023..YR2023..y'])
gdp_clean = data['X2023..YR2023..x']
life_clean = data['X2023..YR2023..y']
log_gdp = np.log(gdp_clean)

print("=== Descriptive Statistics: Raw GDP per capita ===")
print(gdp_clean.describe())

print("\n=== Skewness & Kurtosis: Raw GDP ===")
print("Skewness:", stats.skew(gdp_clean))
print("Kurtosis:", stats.kurtosis(gdp_clean, fisher=False))  # fisher=False matches R's 'moments' package

print("\n=== Skewness & Kurtosis: Log GDP ===")
print("Skewness:", stats.skew(log_gdp))
print("Kurtosis:", stats.kurtosis(log_gdp, fisher=False))

print("\n=== Correlation with Life Expectancy ===")
print("Raw GDP vs Life Expectancy:", np.corrcoef(gdp_clean, life_clean)[0, 1])
print("Log GDP vs Life Expectancy:", np.corrcoef(log_gdp, life_clean)[0, 1])

print("\n=== Linear Regression: Life Expectancy ~ Log(GDP) ===")
slope, intercept, r_value, p_value, std_err = stats.linregress(log_gdp, life_clean)
print(f"Intercept: {intercept:.4f}")
print(f"Slope: {slope:.4f}")
print(f"R-squared: {r_value**2:.4f}")
print(f"P-value: {p_value:.2e}")
```
